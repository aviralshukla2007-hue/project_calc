Data Visualisation is the process of representing the data in the visual and the graphical form, such as bar charts, line graphs, pie charts, histograms, and
scatter plots.

**Reasons to use the Data Visualisation**


*   Understand the large amounts of data.
*   Compare the different categories.
*   To identify the trends and the patterns.
*   To communicate and convey the results directly.
*   To support the data-driven-decision-making.







**Comparision of the Visualization Libraries**

*   Matplotlib: General-purpose plotting
*   Seaborn: Statistical Visualisation
*   Plotly: Interactive Visualisation
*   Pandas: Data Analysis and Quick Plotting.









```python
import matplotlib.pyplot as plt

subjects = ["Python","DBMS","OS","CN"]
marks = [30,25,20,25]

plt.pie(
    marks,
    labels = subjects,
    autopct = "%1.1f%%",
    startangle=90,
    colors=["skyblue","orange","lightgreen","pink"],
    explode = (0.1,0,0,0),
    shadow = True
)

plt.title("Subject-wise Marks distribution")
plt.axis("equal")
plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Aman","Riya","Neha","Rahul"]
marks = [75,80,65,90]

plt.bar(
    students,
    marks,
    color="skyblue",
    edgecolor="black",
    width=0.6,
    label="Student Marks",
)

plt.title("Student Marks Comparision")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.ylim(0,100)
plt.legend()
plt.grid(axis="y",linestyle="--",alpha=0.5)
```
