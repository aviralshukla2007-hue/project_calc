```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
data ={
    "Python":[80,75,90,85],
    "DBMS":[70,80,85,90],
    "OS":[60,75,80,85],
    "CN":[65,70,75,80]
}
df = pd.DataFrame(
    data,
    index=["Student1","Student2","Student3","Student4"]
)
print(df)
sns.heatmap(df,
            annot=True,
            cmap="viridis",
            linewidths=2,
            linecolor="black",
            square=False,
            fmt=".2f"
            )
plt.title("Student Marks Heatmap")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Branch":["CSE","CSE","CSE","AIML","AIML","AIML","IoT","IoT","IoT","CSE"],
    "Marks":[80,75,75,60,85,90,62,64,65,59]
}

df = pd.DataFrame(data)
print(df)

sns.stripplot(x="Branch",
              y="Marks",
              data=df,
              jitter=0.2,
              palette=["blue","green","pink"],
              hue="Branch",
              legend=True,
              size=12
              )

plt.title("Strip plot of the student marks")
plt.xlabel("Branch")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

data = {
    "day":["Mon","Mon","Tue","Tue","Tue","Wed","Wed","Thu","Thu","Fri"],
    "gender":["Male","Female","Male","Female","Male","Female","Male","Female","Male","Female"]
}
df = pd.DataFrame(data)
print(df)

sns.countplot(x="day",
              hue="gender",
              data=df,
              palette=["green","red"],
              )
plt.title("Count plot of the gender",fontsize=15)
plt.xlabel("Days")
plt.ylabel("Gender")
plt.show()
```
